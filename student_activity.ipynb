```python
from google.colab import drive
import numpy as np
import pandas as pd
```

```python
data.isnull().sum()
```

```python
data["Age"]=data["Age"].fillna(data["Age"].median())
```

```python
data["Study_Hours_Per_Day"]=data["Study_Hours_Per_Day"].fillna(data["Study_Hours_Per_Day"].median())
```

```python
data["Attendance_Percent"]=pd.to_numeric(data["Attendance_Percent"], errors="coerce")
```

```python
data["Attendance_Percent"]=data["Attendance_Percent"].fillna(data["Attendance_Percent"].median())
```

```python
data["Parental_Support"].unique()
```

```python
data["Parental_Support"]=data["Parental_Support"].replace({
    "HIGN":"High",
    "MEDIUM":"Medium",
    "medium":"Medium",
    "LOW":"Low",
    "low":"Low"
})
```

```python
data["Gender"].unique()
```

```python
data["Gender"] = (
    data["Gender"]
    .str.strip()
    .str.lower()
    .map({"female": "Female", "male": "Male","other":"other"})
)
```

```python
data["Extra_Classes"].unique()
```

```python
data["Extra_Classes"] = (
    data["Extra_Classes"]
    .str.strip()
    .str.lower()
    .map({"yes": "Yes", "no": "No",})
)
```

```python
data["Pass_Fail"].unique()
```

```python
data["Pass_Fail"] = (
    data["Pass_Fail"]
    .str.strip()
    .str.lower()
    .map({"pass": "Pass","fail": "Fail"}))
```

```python
data["Internet_Access"]=(data["Internet_Access"].str.strip().str.lower().map({
    "yes":"Yes",
    "no":"No"
}))
```

```python
data["Previous_Exam_Marks"]=data["Previous_Exam_Marks"].fillna(data["Previous_Exam_Marks"].median())
data["Assignment_Score"]=data["Assignment_Score"].fillna(data["Assignment_Score"].median())
data["Screen_Time_Hours"]=data["Screen_Time_Hours"].fillna(data["Screen_Time_Hours"].median())
data["Sleep_Hours"]=data["Sleep_Hours"].fillna(data["Sleep_Hours"].median())
data["Internet_Access"]=data["Internet_Access"].fillna(data["Internet_Access"].mode()[0])
data["Extra_Classes"]=data["Extra_Classes"].fillna(data["Extra_Classes"].mode()[0])
data["Parental_Support"]=data["Parental_Support"].fillna(data["Parental_Support"].mode()[0])
data["Pass_Fail"]=data["Pass_Fail"].fillna(data["Pass_Fail"].mode()[0])
data["Final_Exam_Marks"]=data["Final_Exam_Marks"].fillna(data["Final_Exam_Marks"].median())
```

```python
data = data.drop(columns=["Student_ID", "Age"], errors="ignore")
```

```python
data=pd.get_dummies(data, columns=["Parental_Support","Internet_Access","Extra_Classes","Gender"],dtype=int)
```

```python
data.head()
```

```python
# Numerical columns only
numeric_columns = data.select_dtypes(include="number").columns

# Remove target columns if you are using them for prediction
# Change/remove these according to your ML task
exclude_columns = ["Pass_Fail","Final_Exam_Marks"]

numeric_columns = [
    col for col in numeric_columns
    if col not in exclude_columns
]

# Create a mask: initially keep all rows
mask = pd.Series(True, index=data.index)

# IQR method
for column in numeric_columns:

    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Keep values inside the normal range
    mask &= (
        (data[column] >= lower_bound) &
        (data[column] <= upper_bound)
    )

# Remove all rows containing outliers
data = data[mask].copy()

# Reset index
data.reset_index(drop=True, inplace=True)

print("Data shape after removing outliers:", data.shape)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,LogisticRegression
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.svm import SVR
```

```python
x=data.drop(columns=["Pass_Fail","Final_Exam_Marks"],axis=True)
y_classifacation=data["Pass_Fail"]
y_regeration=data["Final_Exam_Marks"]
```

```python
X_train, X_test, y_class_train, y_class_test = train_test_split(
    x,
    y_classifacation,
    test_size=0.2,
    random_state=42,
    stratify=y_classifacation
)

X_train_reg = X_train
X_test_reg = X_test

y_reg_train = y_regeration.loc[X_train.index]
y_reg_test = y_regeration.loc[X_test.index]
```

```python
classification_models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "KNN": KNeighborsClassifier(),
    "Random Forest": RandomForestClassifier(random_state=42),
    "SVM": SVC()
}
classification_results = []
for name,model in classification_models.items():
  model.fit(X_train,y_class_train)
  y_predict=model.predict(X_test)

  classification_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_class_test, y_predict),
        "Precision": precision_score(
            y_class_test,
            y_predict,
            pos_label="Pass",
            zero_division=0
        ),
        "Recall": recall_score(
            y_class_test,
            y_predict,
            pos_label="Pass",
            zero_division=0
        ),
        "F1 Score": f1_score(
            y_class_test,
            y_predict,
            pos_label="Pass",
            zero_division=0
        )
    })

classification_results = pd.DataFrame(classification_results)

print(classification_results)
```

```python
regression_models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "KNN": KNeighborsRegressor(),
    "Random Forest": RandomForestRegressor(random_state=42),
    "SVR": SVR()
}

regression_results = []

for name, model in regression_models.items():

    model.fit(X_train, y_reg_train)

    y_predict = model.predict(X_test)

    regression_results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_reg_test, y_predict),
        "RMSE": np.sqrt(mean_squared_error(y_reg_test, y_predict)),
        "R2 Score": r2_score(y_reg_test, y_predict)
    })

regression_results = pd.DataFrame(regression_results)

print(regression_results)
```
