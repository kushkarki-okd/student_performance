```python
data["Age"]=data["Age"].replace(-1,np.nan)
```

```python
data["Age"]=data["Age"].fillna(data["Age"].median())
```

```python
data["Age"]
```

```python
data["Study_Hours"]=pd.to_numeric(data["Study_Hours"],errors="coerce")
```

```python
data["Study_Hours"]=data["Study_Hours"].fillna(data["Study_Hours"].mean())
```

```python
data["Attendance"]=pd.to_numeric(data["Attendance"],errors="coerce")
data["Attendance"]=data["Attendance"].fillna(data["Attendance"].mean())
```

```python
data["Final_Grade"]=pd.to_numeric(data["Final_Grade"],errors="coerce")
data["Final_Grade"]=data["Final_Grade"].fillna(data["Final_Grade"].mean())
```

```python
data.isnull().sum()
```

```python
data["Sleep_Hours"]=pd.to_numeric(data["Sleep_Hours"],errors='coerce')
data["Sleep_Hours"]=data["Sleep_Hours"].fillna(data["Sleep_Hours"].mean())
```

```python
data.head()
```

```python
data["Gender"].unique()
```

```python
data["Gender"]=data["Gender"].replace({
    "M":"Male",
    "male":"Male",
    "F":"Female"
})
```

```python
data["Internet_Access"].unique()
```

```python
data["Internet_Access"]=data["Internet_Access"].replace({
    "YES":"yes",
    "Yes":"yes",

})
```

```python
data.duplicated().sum()
```

```python
data = data.drop_duplicates()
```

```python
data["Gender"]=data["Gender"].map({
    "Male":1,
    "Female":0
})
```

```python
data=pd.get_dummies(data,columns=["Internet_Access"],dtype=int)
```

```python
data=data.drop(["Student_ID","Name"],axis=1)
```

```python
data["Assignments"]=data["Assignments"].replace({"Nine":9})
```
